```python
import pandas as pd
from pathlib import Path

DATA = Path("data")

def raw(name, **kw):
    return pd.read_csv(DATA  / name, dtype=str, keep_default_na=False, na_filter=False, **kw)
```

```python
h1 = raw("trips_2025_H1.csv")
print("h1:", h1.shape)
h1.head()
```
